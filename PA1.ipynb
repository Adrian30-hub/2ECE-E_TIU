##### A. WORD ROTATION PROBLEM

```python
def rotate_word(text):
    if len(text) != 0:
        a = text[0]
        b = text[1:]
        c = b + a
        print(c)

rotate_word(str(input("Enter your word:")))
```

##### C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
    first, *middle, last = items
    return (last,  *middle, first)

swap_bookends(input("input a list separated by a comma:").split(","))
```
