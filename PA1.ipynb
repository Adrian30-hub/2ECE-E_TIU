##### A. WORD ROTATION PROBLEM

```python
def rotate_word(text):
    # Check if the input string is not empty to avoid error
    if len(text) != 0:
        # Extract First Character
        a = text[0]
        # Extract remaining elements from 1 to end
        b = text[1:]
        # Concatenate a and b
        c = b + a
        # Print output
        print(c)

# Call function
rotate_word(str(input("Enter your word:")))
```

##### B. USERNAME BUILDER PROBLEM

```python
# Get input from user
a = str(input("First Name:"))
b = str(input("Last Name:"))

def make_username(fName, lName):
    # Concatenate first name, "." and last name
    c = fName + "." + lName
    # Make all cases in lower
    c = c.lower()
    # Print output
    print(c)

# Call function
make_username(a,b)
```

##### C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
    # Unpack into first element, middle list, and last element
    first, *middle, last = items
    # Return with the first and last elements swapped
    return (last,  *middle, first)

# Get user input and split commas and pass to the function
swap_bookends(input("input a list separated by a comma:").split(","))
```
